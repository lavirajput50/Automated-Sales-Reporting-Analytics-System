```python
# Missing Value Validation for each columns 
# And also clean data percentage and missing data percentage 
def missing_value(df):
    print("=======Complete Missing Values Report=======")
    report = pd.DataFrame({
        'Missing_Count': df.isnull().sum(),
        'Missing_%': (df.isnull().sum() / len(df)) * 100,
        'Clean_%': 100 - ((df.isnull().sum() / len(df)) * 100)
    })
    return report
a=missing_value(dt)
print(a)
    
```

```python
# Duplicate And Uniquess Validation For Order_id
def duplicated_data(data,col_name):
    print(f"======Duplicate Validation {col_name}===== ")
    duplicated_data=data[col_name].duplicated().sum()
    return (duplicated_data)

result=duplicated_data(dt,'Order_ID')
print(result)
```

data cleaning in Cuisine_Type,   Weather_Condition, Complaint_Flag, Refund_Flag, City

```python
dt2['Cuisine_Type']=dt2['Cuisine_Type'].str.strip().str.title() # Remove extra space and convert into title'
dt2['Weather_Condition']=dt2['Weather_Condition'].str.strip().str.title() # Remove extra space and convert into title  
```

```python
dt2['Cuisine_Type']=dt2['Cuisine_Type'].astype(str).replace(['','-','Nan','null','Null','N/A'],"Unknown") # replace all '-','Nan',etc into "Unknown"
dt2['Cuisine_Type'].unique() 
```

```python
dt2['Weather_Condition']=dt2['Weather_Condition'].astype(str).replace(['','-','Nan','null','Null',
'N/A'],"Unknown") # replace all '-','Nan',etc into "Unknown"
dt2['Weather_Condition'].unique() 
```

```python
# For Complaint_Flag
# Replace - ("0","False","No") with 0 and ("Yes","True","1") with 1
dt2['Complaint_Flag']=dt2['Complaint_Flag'].replace({"False":0,"0":0,"No":0,"Yes":1,"True":1,"1":1}).astype(int)
dt2['Complaint_Flag'].unique() # after dtype conversion and replaced values 
```

```python
# For Refund_Flag
dt2['Refund_Flag']=dt2['Refund_Flag'].replace({'False':0,'No':0,' ':np.nan,'-':np.nan,'Yes':1, 'unknown':np.nan,'True':1}).fillna(0).astype(int) # Standardized boolean values (False/No → 0, True/Yes → 1), converted blank/- values to NaN, and filled missing values with 0.
dt2['Refund_Flag'].unique() # after complete all process
```

```python
# For City columns 
dt2['City']=dt2['City'].str.strip() # Remove extra spacing from city column values

dt2['City']=dt2['City'].str.title() # convert inot title case
```

```python
# Replace '-','','Null',etc into 'Unknown' for city column 
dt2['City']=dt2['City'].replace(['','-','Null','Nan','N/A'],"Unknown").astype(str)
```

```python
 # City Name Standardization and Validation for city column
dt2['City']=dt2['City'].replace({'Sydney, Au':'Sydney', 'Sg': 'Singapore', 'London, Uk':'London', 'New York City': 'New York', 'Ny':'New York','Newyork':'New York'}).astype(str)
dt2["City"].unique()
```

Data cleaning & validation for numerical columns - Customer_Age, Delivery_Distance_KM, Delivery_Time_Min, Order_Value_USD, Item_Count, Customer_Rating, Revenue_USD, Profit_USD, Month, Quarter, Demand_Score, Churn_Risk


```python
dt2["Delivery_Distance_KM"]=dt2["Delivery_Distance_KM"].str.replace(" km", "",regex=True).replace({"-":np.nan,"unknown":np.nan,"Unknown":np.nan,"":np.nan," ":np.nan}).astype(float)
```

```python
dt2 = dt2.dropna(subset=["Delivery_Distance_KM"]) # Remove all missing values from 'Delivery_Distance_KM' 
dt2['Delivery_Distance_KM'].unique()
```

```python
# For Delivery_Time_Min

# remove suffix 'm','unknown' 
dt2['Delivery_Time_Min'] = (
    dt2['Delivery_Time_Min']
    .replace([' ','-','nan','unknown','Unknown'], np.nan)   # messy values ko NaN
    .astype(str)                                 # sabko string banao
    .str.replace(r"(min|in)$", "", regex=True)   # suffix 'min' ya 'in' hatao
)
```

```python
dt2["Delivery_Time_Min"] = dt2["Delivery_Time_Min"].replace("nan", np.nan) # Again 'nan' value convert into NaN
dt2=dt2.dropna(subset=['Delivery_Time_Min']) # Now, Remove all missing values or NaN inside the columns
display(dt2['Delivery_Time_Min'].unique()) # after complete all process, again check the unquies values 
```

```python
# For Order_Value_USD
dt2['Order_Value_USD']=(dt2['Order_Value_USD']
                        .replace([" ","","-","unknown","Unknown"],np.nan) 
                        .astype(str).replace(r"(USD|usd)$", "",regex=True)
                        )
```

```python
dt2=dt2.dropna(subset=['Order_Value_USD']) # Remove missing values 
dt2['Order_Value_USD'].unique()  
```

```python
# For Item_Count
dt2['Item_Count'].unique() 
# Item_Count column is already clean, with no null or missing values.
```

```python
# For Customer_Rating 
dt2['Customer_Rating']=dt2['Customer_Rating'].replace([' ','','-','unknown'],np.nan).astype(float)
dt2=dt2.dropna(subset=['Customer_Rating'])
dt2['Customer_Rating'].unique()
```

```python
# For Revenue_USD Column
# Step 1: Replace blanks and unknowns with NaN
dt2['Revenue_USD'] = dt2['Revenue_USD'].replace(['','unknown','Unknown','Uknown','-',' '], np.nan)

# Step 2: Remove dollar sign and 'USD' text
dt2['Revenue_USD'] = dt2['Revenue_USD'].replace(r'\$', '', regex=True)
dt2['Revenue_USD'] = dt2['Revenue_USD'].replace(r'USD', '', regex=True)

# Step 3: Strip extra spaces
dt2['Revenue_USD'] = dt2['Revenue_USD'].str.strip()

```

```python
# convert Revenue_USD column into float
dt2['Revenue_USD']=dt2['Revenue_USD'].astype(float)
```

```python
# For Profit_USD column
dt2['Profit_USD']=dt2['Profit_USD'].replace(['',' ','-','unknown','Unknown'],np.nan) # replace blanks,'-','Unknown' with NaN
# Remove $ and USD sign
dt2['Profit_USD']=dt2['Profit_USD'].replace(r'\$','',regex=True)
dt2['Profit_USD']=dt2['Profit_USD'].replace(r'USD','',regex=True)

# convert into float and positive
dt2['Profit_USD']=dt2['Profit_USD'].astype(float)
dt2['Profit_USD']
```
