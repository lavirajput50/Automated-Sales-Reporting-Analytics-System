```python
# Shape, columns, data types 
def check_overview(data):
    print('Shape of Dataset-',data.shape,end='')
    print('\nColoumns of datatset -',data.columns,end='')
    print('\nData types of datasets columns-',data.dtypes,end='')
check_overview(dt)
```

```python
# Check Missing Values in Data
def missing_values(data):
    print("\n Missing Values Of Each Columns-",data.isnull().sum())
    print("\n Total Missing Values In Dataset-",data.isnull().sum().sum())
missing_values(dt)
```

```python
# check categorical columns 
# check numerical columns 
# statistics of numerical 
def check_summary(data):
    num_col=data.select_dtypes(include='number').columns
    cat_col=data.select_dtypes(include='object').columns
    stats=data[num_col].describe()
    print("\nCheck numerical columns in dataset\n",num_col)
    print("\n Check categorical columns in dataset\n",cat_col)
    print('\nStatistics of numberical columns\n',stats)

check_summary(dt)
```

## Overview:
1. The dataset contains 50,000 records and 21 columns related to food delivery operations. Initial profiling shows that the dataset contains significant data-quality issues. Although only two columns are recognized by Pandas as numerical (Item_Count and Quarter), several other business-numerical columns such as Revenue_USD, Profit_USD, Order_Value_USD, Customer_Age, Delivery_Time_Min, Delivery_Distance_KM, Customer_Rating, Demand_Score, and Churn_Risk are stored as object data types due to inconsistent formatting, units, currency symbols, and invalid text values.
2.  The dataset contains 757 duplicate records and 757 duplicate Order IDs. Multiple columns contain missing values, while additional missing-like values such as -, blank strings, unknown, null, and n/a are also present. Categorical fields contain inconsistent capitalization, spacing, abbreviations, and naming conventions. Suspicious values such as a customer age of 999 and a customer rating of 10 were also identified.
3. Therefore, the dataset requires data cleaning, datatype correction, missing-value standardization, categorical normalization, duplicate handling, and validation of invalid/outlier values before performing reliable EDA and business analysis.
